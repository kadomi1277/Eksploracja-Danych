W Mercury zmiana widżetu uruchamia ponownie **komórki poniżej** tej, w której widżet powstał.

```
import mercury as mr
import sales_logic as L

conn = L.get_connection()
products = L.get_products(conn)
```

```
chosen = mr.MultiSelect(label="Produkty", value=products, choices=products)
dates = mr.DateRange(label="Zakres dat", value=["2026-09-01", "2026-11-30"])
```

```
df = L.get_sales(conn, chosen.value, *dates.value)

mr.Indicator([
    mr.Indicator(value=f"{len(df):,}".replace(",", " "), label="Transakcje"),
    mr.Indicator(value=f"{df.revenue.sum():,.0f} zł".replace(",", " "), label="Przychód", variant="success"),
])
```

```
L.fig_daily(df).show()
```

```
L.fig_products(df).show()
L.fig_map(df).show()
```
